# PySpark RDD — Food Data Analytics

Demonstrates core RDD transformation operations — `map`, `filter`, and `distinct` — on a CSV food dataset.

**Dataset schema:** `FoodName, Category, Color, Calories`

---
## 1. Environment Setup

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.master("local").appName("Colab PySpark").getOrCreate()
print(spark)
```

```python
sc = spark.sparkContext
print(sc)
```

---
## 2. Load Dataset

```python
food_rdd = sc.textFile("/content/sample_data/food_data.csv")
food_rdd.collect()
```

---
## 3. Map Transformations

```python
# Q1. Extract only the food names from each row
food_names = food_rdd.map(lambda x: x.split(",")).map(lambda x: x[0])
food_names.collect()
```

```python
# Q2. Convert each food name to lowercase
food_names_lowercase = food_names.map(lambda x: x.lower())
food_names_lowercase.collect()
```

```python
# Q3. Convert each food name to uppercase
food_names_uppercase = food_names.map(lambda x: x.upper())
food_names_uppercase.collect()
```

```python
# Q4. Extract calorie values and cast to integer
calories = food_rdd.map(lambda x: x.split(",")).map(lambda x: int(x[3]))
calories.collect()
```

```python
# Q5. Create (FoodName, Calories) key-value pairs
food_name_calories = food_rdd.map(lambda x: x.split(",")).map(lambda x: (x[0], int(x[3])))
food_name_calories.collect()
```

```python
# Q6. Append the suffix '_FOOD' to every food name
food_names_suffixed = food_rdd.map(lambda x: x.split(",")).map(lambda x: x[0] + "_FOOD")
food_names_suffixed.collect()
```

```python
# Q7. Create distinct (Category, 1) key-value pairs
distinct_categories = food_rdd.map(lambda x: x.split(",")).map(lambda x: (x[1], 1)).distinct()
distinct_categories.collect()
```

```python
# Q8. Multiply each calorie value by 2
calories_doubled = calories.map(lambda x: x * 2)
calories_doubled.collect()
```

```python
# Q9. Create (Color, Calories) key-value pairs
color_calories = food_rdd.map(lambda x: x.split(",")).map(lambda x: (x[2], int(x[3])))
color_calories.collect()
```

```python
# Q10. Add 10 calories to each food item
calories_plus_ten = food_rdd.map(lambda x: x.split(",")).map(lambda x: (x[0], int(x[3]) + 10))
calories_plus_ten.collect()
```

---
## 4. Filter Transformations

```python
# Q11. Filter food items where calories > 100
high_calorie_foods = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[3]))
    .filter(lambda x: int(x[1]) > 100)
)
high_calorie_foods.collect()
```

```python
# Q12. Filter food items from the 'Fruit' category
fruit_items = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[1]))
    .filter(lambda x: x[1] == "Fruit")
)
fruit_items.collect()
```

```python
# Q13. Filter food items where color is 'Green'
green_food_items = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[2]))
    .filter(lambda x: x[1] == "Green")
)
green_food_items.collect()
```

```python
# Q14. Filter food names that start with the letter 'B'
food_names_starting_with_b = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: x[0])
    .filter(lambda x: x.startswith("B"))
)
food_names_starting_with_b.collect()
```

```python
# Q15. Filter food items where calories < 80
low_calorie_foods = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[3]))
    .filter(lambda x: int(x[1]) < 80)
)
low_calorie_foods.collect()
```

```python
# Q16. Filter food items from the 'Dairy' category
dairy_items = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[1]))
    .filter(lambda x: x[1] == "Dairy")
)
dairy_items.collect()
```

```python
# Q17. Filter food items where color is 'White'
white_food_items = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[2]))
    .filter(lambda x: x[1] == "White")
)
white_food_items.collect()
```

```python
# Q18. Filter food items where calories are between 90 and 120 (exclusive)
mid_calorie_foods = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[3]))
    .filter(lambda x: 90 < int(x[1]) < 120)
)
mid_calorie_foods.collect()
```

```python
# Q19. Filter food items from the 'Snack' category
snack_items = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: (x[0], x[1]))
    .filter(lambda x: x[1] == "Snack")
)
snack_items.collect()
```

```python
# Q20. Filter food items where name length > 6 characters
long_name_foods = (
    food_rdd
    .map(lambda x: x.split(","))
    .map(lambda x: x[0])
    .filter(lambda x: len(x) > 6)
)
long_name_foods.collect()
```
